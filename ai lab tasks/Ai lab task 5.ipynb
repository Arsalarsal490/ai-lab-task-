```python
#Task 1
import math

def minimax(curDepth, nodeIndex, maxTurn, scores, targetDepth):
    # Base case: leaf node reached
    if curDepth == targetDepth:
        return scores[nodeIndex]

    if maxTurn:
        return max(
            minimax(curDepth + 1, nodeIndex * 2,
                    False, scores, targetDepth),
            minimax(curDepth + 1, nodeIndex * 2 + 1,
                    False, scores, targetDepth)
        )
    else:
        return min(
            minimax(curDepth + 1, nodeIndex * 2,
                    True, scores, targetDepth),
            minimax(curDepth + 1, nodeIndex * 2 + 1,
                    True, scores, targetDepth)
        )


# Given leaf values
scores = [19, 3, 9, 2, 1, 5, 19]

# Pad to complete binary tree
treeDepth = math.ceil(math.log2(len(scores)))
scores += [0] * (2**treeDepth - len(scores))

print("Leaf values:", scores)
print("Tree depth:", treeDepth)

result = minimax(0, 0, True, scores, treeDepth)

print("Optimal Minimax value:", result)
```

```python
#Task 2
def minimax(d, i, is_max, scores, target, count):
    if d == target:
        count[0] += 1
        return scores[i]

    a = minimax(d+1, i*2, not is_max,
                scores, target, count)
    b = minimax(d+1, i*2+1, not is_max,
                scores, target, count)

    return max(a, b) if is_max else min(a, b)


def alphabeta(d, i, is_max, scores, target,
              alpha, beta, count):
    if d == target:
        count[0] += 1
        return scores[i]

    if is_max:
        value = float("-inf")
        for j in range(2):
            value = max(value, alphabeta(
                d+1, i*2+j, False, scores, target,
                alpha, beta, count))
            alpha = max(alpha, value)
            if alpha >= beta:
                break
        return value

    value = float("inf")
    for j in range(2):
        value = min(value, alphabeta(
            d+1, i*2+j, True, scores, target,
            alpha, beta, count))
        beta = min(beta, value)
        if alpha >= beta:
            break
    return value


scores = [19, 3, 9, 2, 1, 5, 19, 0]
depth = 3

c1 = [0]
v1 = minimax(0, 0, True, scores, depth, c1)

c2 = [0]
v2 = alphabeta(
    0, 0, True, scores, depth,
    float("-inf"), float("inf"), c2
)

print("Minimax value:", v1)
print("Minimax leaves:", c1[0])
print("Alpha-Beta value:", v2)
print("Alpha-Beta leaves:", c2[0])

```
