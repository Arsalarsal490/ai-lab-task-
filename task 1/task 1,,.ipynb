```python
def check_application(score, documents_complete, attendance, interview_score):
    # Original conditions
    if score < 70:
        return "Fail"
    if not documents_complete:
        return "Fail"

    # Two additional transparent conditions
    if attendance < 75:
        return "Fail"
    if interview_score < 60:
        return "Fail"

    return "Pass"


test_cases = [
    {
        "name": "1. Pass",
        "input": {
            "score": 80,
            "documents_complete": True,
            "attendance": 85,
            "interview_score": 75,
        },
        "expected": "Pass",
        "reason": "All four conditions are satisfied."
    },
    {
        "name": "2. Attendance condition fails",
        "input": {
            "score": 80,
            "documents_complete": True,
            "attendance": 70,
            "interview_score": 75,
        },
        "expected": "Fail",
        "reason": "Attendance is 70%, below the required 75%."
    },
    {
        "name": "3. Interview condition fails",
        "input": {
            "score": 80,
            "documents_complete": True,
            "attendance": 85,
            "interview_score": 50,
        },
        "expected": "Fail",
        "reason": "Interview score is 50, below the required 60."
    },
    {
        "name": "4. Threshold boundary",
        "input": {
            "score": 70,
            "documents_complete": True,
            "attendance": 75,
            "interview_score": 60,
        },
        "expected": "Pass",
        "reason": "Each numeric value is exactly at its minimum threshold, and >= means the boundary is accepted."
    },
]

print("Thresholds are synthetic lab assumptions.")
print("Rules: score >= 70, documents complete, attendance >= 75%, interview score >= 60.")
print()

for case in test_cases:
    data = case["input"]
    actual = check_application(**data)
    print(case["name"])
    print("Input:", data)
    print("Expected result:", case["expected"])
    print("Actual result:", actual)
    print("Reason:", case["reason"])
    print("Test:", "PASS" if actual == case["expected"] else "FAIL")
    print("-" * 60)
```
